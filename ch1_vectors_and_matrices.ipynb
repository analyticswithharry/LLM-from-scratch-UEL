#### 1. LLM Architecture: From Mathematics to Transformers

Large Language Models (LLMs) are advanced artificial intelligence systems designed to understand and generate human language. They can perform tasks such as answering questions, summarizing documents, translating languages, writing code, and generating creative content. Modern LLMs, including GPT-style models, are primarily built using a deep learning architecture called the **Transformer**.

The Transformer architecture was introduced in the research paper *Attention Is All You Need* in 2017. It uses a mechanism called **self-attention** to learn relationships between tokens in a sequence, allowing the model to determine which parts of the input are most relevant when processing each token.

By combining theoretical foundations with hands-on coding, this project aims to build a clear understanding of how Transformer layers process information, how model parameters are learned during training, and how a trained language model generates text one token at a time.

##### 1.2 Input Representation

Token embeddings represent token identities. Positional information represents token order.

$$
X = E_{\text{token}} + E_{\text{position}}
$$

##### 1.3 Query, Key, and Value Projections

The input matrix is projected into three representations using trainable weight matrices.

$$
Q = X W_{Q}
$$

$$
K = X W_{K}
$$

$$
V = X W_{V}
$$

Where:

- $X$ is the input representation matrix.
- $Q$ is the query matrix.
- $K$ is the key matrix.
- $V$ is the value matrix.
- $W_{Q}$, $W_{K}$, and $W_{V}$ are trainable projection matrices.


##### 1.4 Attention Scores

The dot product between queries and keys measures the compatibility between tokens.

$$
S = Q K^{\top}
$$

Here, $K^{\top}$ denotes the transpose of the key matrix.

##### 1.5 Scaled Attention Scores

The scores are scaled by the square root of the key dimension.

$$
S_{\text{scaled}}
=
\frac{Q K^{\top}}{\sqrt{d_{k}}}
$$

Where $d_{k}$ is the dimension of each key vector.

##### 1.6 Attention Weights

Softmax converts the scaled scores into normalized attention weights.

$$
A =
\operatorname{softmax}
\left(
\frac{Q K^{\top}}{\sqrt{d_{k}}}
\right)
$$

For a decoder-only language model such as GPT, this operation also uses a causal mask to prevent each token from attending to future tokens.


##### 1.7 Attention Output

The attention weights are multiplied by the value matrix.

$$
O_{\text{attention}} = A V
$$

The complete scaled dot-product attention formula is:

$$
\operatorname{Attention}(Q,K,V)
=
\operatorname{softmax}
\left(
\frac{Q K^{\top}}{\sqrt{d_{k}}}
\right)V
$$


##### 1.8 Multi-Head Attention

Each attention head learns different relationships between tokens.

$$
\operatorname{head}_{i}
=
\operatorname{Attention}
\left(
X W_{Q}^{(i)},
X W_{K}^{(i)},
X W_{V}^{(i)}
\right)
$$

The head outputs are concatenated and projected:

$$
\operatorname{MultiHead}(X)
=
\operatorname{Concat}
\left(
\operatorname{head}_{1},
\ldots,
\operatorname{head}_{h}
\right)W_{O}
$$

Where $h$ is the number of attention heads and $W_{O}$ is the output projection matrix.

##### 1.9 Residual Connections and Layer Normalization

A common pre-layer-normalization Transformer block first normalizes the input before attention:

$$
H = X + \operatorname{MultiHead}(\operatorname{LayerNorm}(X))
$$

Then it applies the feed-forward network with another residual connection:

$$
Y = H + \operatorname{FFN}(\operatorname{LayerNorm}(H))
$$

Here, $H$ and $Y$ are intermediate representations passed through the Transformer block.

##### 1.10 Feed-Forward Network

The feed-forward network applies two linear transformations with a nonlinear activation function between them.

$$
\operatorname{FFN}(x)
=
\operatorname{Activation}
\left(
x W_{1} + b_{1}
\right)W_{2} + b_{2}
$$

The activation function may be GELU or another suitable nonlinear function.

##### 1.11 Next-Token Prediction

The final hidden representation is projected into vocabulary-sized logits.

$$
z = h W_{\text{vocab}} + b_{\text{vocab}}
$$

Softmax converts the logits into probabilities:

$$
P(\text{token}\mid\text{context})
=
\operatorname{softmax}(z)
$$

The model selects the next token, appends it to the context, and repeats the process to generate text.

##### 1.12 Notation Reference

| Symbol | Meaning |
|---|---|
| $X$ | Input representation matrix |
| $Q$ | Query matrix |
| $K$ | Key matrix |
| $V$ | Value matrix |
| $W_{Q}$ | Query projection weights |
| $W_{K}$ | Key projection weights |
| $W_{V}$ | Value projection weights |
| $K^{\top}$ | Transpose of the key matrix |

| $A$ | Attention weight matrix |
| $W_{O}$ | Output projection weights |
| $W_{1}, W_{2}$ | Feed-forward network weights |
| $h$ | Final hidden representation |
| $z$ | Vocabulary logits |

#### Matrix–Vector Multiplication Formula

For a matrix \(M\) and a vector \(v\), matrix–vector multiplication is defined as:

$$
y_i = \sum_{j=1}^{n} M_{ij}v_j
$$

Where:

- $M_{ij}$ is the element in row $i$, column $j$ of the matrix.
- $v_j$ is the $j$-th element of the vector.
- $y_i$ is the $i$-th element of the output vector.
- $n$ is the number of columns in the matrix.

For a matrix with 3 rows and 2 columns:

$$

\end{bmatrix}
$$

The resulting vector has shape \((3,)\).

### Autoregressive Language Modelling Pipeline

- Tokenization — poem to token IDs.
- Embeddings — token IDs to numerical vectors.
- Self-attention — tokens compare and combine information.
- Causal masking — tokens cannot look at future tokens.
- Output layer — scores possible next tokens.

````python
# Vocabulary, token ID
vocab = {
    "The": 0,
    "knight": 1,
    "rides": 2,
    "through": 3,
    "darkness,": 4,
    "seeking": 5,
    "glory": 6,
    "beyond": 7,
    "the": 8,
    "stars.": 9,
    "fire": 10,
    "battle": 11,
    "victory": 12,
    "shadows": 13
}
````

````python
# input poem (10 words)
tokens = [
    "The", "knight", "rides", "through", "darkness,",
    "seeking", "glory", "beyond", "the", "stars."
]
````

````python
token_ids = [vocab[x] for x in tokens]

print("1. Tokens:", tokens)
print("2. Token IDs:", token_ids)
````

````python
# embedding table
np.random.seed(42)
E = np.random.randn(len(vocab), 4)
E
````

````python
X = E[token_ids]

print("\n3. Embedding vectors:")
print(np.round(X, 3))
````

````python
# simplified self-attention
# Q, K, V are X for demonstration only.
Q = K = V = X
Q
````

````python
scores = Q @ K.T / np.sqrt(X.shape[1])
scores
````

````python
# causal mask, each token can attend only to itself
# and earlier tokens, never future tokens.
mask = np.triu(
    np.ones(scores.shape, dtype=bool),
    k=1
)
scores = np.where(mask, -np.inf, scores)
````

````python
mask
````

````python
scores
````

````python
# softmax
exp_scores = np.exp(
    scores - np.max(scores, axis=1, keepdims=True)
)
attention = exp_scores / exp_scores.sum(
    axis=1, keepdims=True
)
````

````python
exp_scores
````

````python
attention
````

````python
# combine value vectors
output = attention @ V

print("\n4. Attention weights:")
print(np.round(attention, 3))

print("\nAttention output for the final token:")
print(np.round(output[-1], 3))
````

````python
# illustrative next-token prediction
# these output weights are random, not trained.
output_weights = np.random.randn(len(vocab), 4)

logits = output_weights @ output[-1]
predicted_id = int(np.argmax(logits))
id_to_token = {i: token for token, i in vocab.items()}

print("\n Next-token scores:")
for token, score in zip(vocab.keys(), logits):
    print(f"{token:>10}: {score:.3f}")
````

````python
print("\nPredicted next token:", id_to_token[predicted_id])
````

#### <h3 align="center">TINY TRANSFORMER USING NUMPY</h3>


````python
class TinyTransformer:
    def __init__(self, embedding_dim=4, seed=42):
        self.embedding_dim = embedding_dim
        self.rng = np.random.default_rng(seed)
        
        # vocabulary and model parameters will be created later
        self.vocab = {}
        self.id_to_token = {}
        self.E = None
        self.output_weights = None
        
    # tokenize text
    def tokenize(self, text):
        return text.split()
        
    # automatically build vocabulary from text
    def fit_vocabulary(self, texts):
        if isinstance(texts, str):
            texts = [texts]
            
        # collect unique tokens in order of first appearance
        unique_tokens = dict.fromkeys(
            token
            for text in texts
            for token in self.tokenize(text)
        )
        
        # assign an integer ID to each token
        self.vocab = {
            token: idx
            for idx, token in enumerate(unique_tokens)
        }

        # reverse mapping: ID, token
        self.id_to_token = {
            idx: token
            for token, idx in self.vocab.items()
        }

        # embeddings for the new vocabulary
        self.E = self.rng.standard_normal(
            (len(self.vocab), self.embedding_dim)
        )

        # weights
        self.output_weights = self.rng.standard_normal(
            (len(self.vocab), self.embedding_dim)
        )
        return self

    # convert text to token IDs
    def encode(self, text):
        tokens = self.tokenize(text)
        unknown = [
            token for token in tokens
            if token not in self.vocab
        ]
        if unknown:
            raise ValueError(
                f"Unknown tokens: {unknown}. "
                "Add more training text using fit_vocabulary()."
            )
        return [self.vocab[token] for token in tokens]

    # convert IDs to text
    def decode(self, token_ids):
        return [
            self.id_to_token[token_id]
            for token_id in token_ids
        ]

    # self-attention
    def self_attention(self, X):
        Q = K = V = X
        scores = Q @ K.T / np.sqrt(self.embedding_dim)
        # prevent looking at future tokens
        mask = np.triu(
            np.ones(scores.shape, dtype=bool),
            k=1
        )
        scores = np.where(mask, -np.inf, scores)
        # softmax
        exp_scores = np.exp(
            scores - np.max(scores, axis=1, keepdims=True)
        )
        attention = exp_scores / exp_scores.sum(
            axis=1, keepdims=True
        )
        return attention @ V, attention

    # predict next token
    def predict(self, text):
        token_ids = self.encode(text)
        
        if not token_ids:
            raise ValueError("Input text cannot be empty.")
        X = self.E[token_ids]

        output, attention = self.self_attention(X)
        final_vector = output[-1]
        logits = self.output_weights @ final_vector
        exp_logits = np.exp(logits - np.max(logits))
        probabilities = exp_logits / exp_logits.sum()
        predicted_id = int(np.argmax(probabilities))
        return {
            "input": text,
            "token_ids": token_ids,
            "attention": attention,
            "predicted_id": predicted_id,
            "predicted_token": self.id_to_token[predicted_id],
            "probability": float(probabilities[predicted_id])
        }

    # vocabulary predicted
    def show_vocabulary(self):
        for token, idx in self.vocab.items():
            print(f"{token!r}: {idx}")
````

#### <h3 align="center">TINY TRANSFORMER USING PYTORCH</h3>


````python
import re
import random
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset
````

````python
# reproducibility
random.seed(42)
np.random.seed(42)
torch.manual_seed(42)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)
````

````python
class TinyTransformer(nn.Module):
    def __init__(
        self,
        texts,
        embedding_dim=64,
        num_heads=4,
        num_layers=2,
        feedforward_dim=128,
        block_size=16,
        dropout=0.1,
    ):
        super().__init__()

        self.embedding_dim = embedding_dim
        self.block_size = block_size

        # tokenize text
        self.tokenize = lambda text: re.findall(r"\w+|[^\w\s]", text.lower())

        # automatically build vocabulary from text
        special_tokens = ["<pad>", "<unk>", "<eos>"]
        all_tokens = [
            token
            for text in texts
            for token in self.tokenize(text)
        ]

        self.tokens = special_tokens + sorted(
            set(all_tokens) - set(special_tokens)
        )

        self.vocab = {
            token: idx
            for idx, token in enumerate(self.tokens)
        }
        self.id_to_token = {
            idx: token
            for token, idx in self.vocab.items()
        }

        self.PAD_ID = self.vocab["<pad>"]
        self.UNK_ID = self.vocab["<unk>"]
        self.EOS_ID = self.vocab["<eos>"]
        self.vocab_size = len(self.vocab)

        # token and position embeddings
        self.token_embedding = nn.Embedding(
            self.vocab_size,
            embedding_dim,
            padding_idx=self.PAD_ID,
        )
        self.position_embedding = nn.Embedding(
            block_size, embedding_dim
        )

        # transformer blocks
        encoder_layer = nn.TransformerEncoderLayer(
            d_model=embedding_dim,
            nhead=num_heads,
            dim_feedforward=feedforward_dim,
            dropout=dropout,
            batch_first=True,
            norm_first=True,
            activation="gelu",
        )

        self.transformer = nn.TransformerEncoder(
            encoder_layer,
            num_layers=num_layers,
            enable_nested_tensor=False,
        )

        # output layer
        self.final_norm = nn.LayerNorm(embedding_dim)
        self.output_layer = nn.Linear(
            embedding_dim, self.vocab_size
        )

    # convert text to token IDs
    def encode(self, text):
        return [
            self.vocab.get(token, self.UNK_ID)
            for token in self.tokenize(text)
        ]

    # convert IDs to text
    def decode(self, token_ids):
        return " ".join(
            self.id_to_token[int(token_id)]
            for token_id in token_ids
            if int(token_id) not in (self.PAD_ID, self.EOS_ID)
        )

    # forward pass through transformer
    def forward(self, input_ids):
        batch_size, seq_len = input_ids.shape

        positions = torch.arange(
            seq_len, device=input_ids.device
        ).unsqueeze(0)

        # combine token and position embeddings
        x = (
            self.token_embedding(input_ids)
            + self.position_embedding(positions)
        )

        # prevent looking at future tokens
        causal_mask = torch.triu(
            torch.ones(
                seq_len,
                seq_len,
                device=input_ids.device,
                dtype=torch.bool,
            ),
            diagonal=1,
        )

        # ignore padding positions
        padding_mask = input_ids.eq(self.PAD_ID)

        # transformer layers
        x = self.transformer(
            x,
            mask=causal_mask,
            src_key_padding_mask=padding_mask,
        )

        # predict next token
        x = self.final_norm(x)
        return self.output_layer(x)
````

````python
class TransformerTrainer:
    def __init__(
        self,
        model,
        learning_rate=0.001,
        weight_decay=0.01,
        batch_size=32,
    ):
        self.model = model.to(device)
        self.batch_size = batch_size

        self.criterion = nn.CrossEntropyLoss(
            ignore_index=-100
        )
        self.optimizer = torch.optim.AdamW(
            self.model.parameters(),
            lr=learning_rate,
            weight_decay=weight_decay,
        )

        # store training history
        self.loss_history = []
        self.accuracy_history = []

    # prepare next-token training examples
    def prepare_data(self, texts):
        token_stream = []

        for text in texts:
            token_stream.extend(self.model.encode(text))
            token_stream.append(self.model.EOS_ID)

        examples = []
        block_size = self.model.block_size

        # sliding windows provide more training examples
        for start in range(0, len(token_stream) - 1, 4):
            chunk = token_stream[start:start + block_size + 1]

            if len(chunk) < 2:
                continue

            x = chunk[:-1]
            y = chunk[1:]

            if len(x) < block_size:
                padding = block_size - len(x)
                x += [self.model.PAD_ID] * padding
                y += [-100] * padding

            examples.append((x, y))

        if not examples:
            raise ValueError(
                "Provide training texts with at least two tokens."
            )

        X = torch.tensor(
            [item[0] for item in examples],
            dtype=torch.long,
        )
        Y = torch.tensor(
            [item[1] for item in examples],
            dtype=torch.long,
        )

        return DataLoader(
            TensorDataset(X, Y),
            batch_size=self.batch_size,
            shuffle=True,
        )

    # evaluate next-token accuracy
    def evaluate_loader(self, loader):
        self.model.eval()

        total_loss = 0.0
        correct = 0
        total = 0
        batches = 0

        with torch.no_grad():
            for batch_x, batch_y in loader:
                batch_x = batch_x.to(device)
                batch_y = batch_y.to(device)

                logits = self.model(batch_x)

                loss = self.criterion(
                    logits.reshape(-1, self.model.vocab_size),
                    batch_y.reshape(-1),
                )

                predictions = logits.argmax(dim=-1)
                valid = batch_y != -100

                total_loss += loss.item()
                correct += (
                    (predictions == batch_y) & valid
                ).sum().item()
                total += valid.sum().item()
                batches += 1

        return {
            "loss": total_loss / max(batches, 1),
            "accuracy": correct / max(total, 1),
        }

    # train model
    def fit(self, texts, epochs=100, verbose=True):
        loader = self.prepare_data(texts)

        self.loss_history = []
        self.accuracy_history = []

        for epoch in range(epochs):
            self.model.train()

            total_loss = 0.0
            correct = 0
            total = 0
            batches = 0

            for batch_x, batch_y in loader:
                batch_x = batch_x.to(device)
                batch_y = batch_y.to(device)

                self.optimizer.zero_grad()

                logits = self.model(batch_x)

                loss = self.criterion(
                    logits.reshape(-1, self.model.vocab_size),
                    batch_y.reshape(-1),
                )

                # automatic backpropagation
                loss.backward()

                # stabilize gradient updates
                torch.nn.utils.clip_grad_norm_(
                    self.model.parameters(),
                    max_norm=1.0,
                )

                self.optimizer.step()

                total_loss += loss.item()
                batches += 1

                predictions = logits.argmax(dim=-1)
                valid = batch_y != -100

                correct += (
                    (predictions == batch_y) & valid
                ).sum().item()
                total += valid.sum().item()

            average_loss = total_loss / max(batches, 1)
            accuracy = correct / max(total, 1)

            self.loss_history.append(average_loss)
            self.accuracy_history.append(accuracy)

            if verbose and (
                epoch == 0
                or (epoch + 1) % 10 == 0
                or epoch == epochs - 1
            ):
                print(
                    f"Epoch {epoch + 1}/{epochs} | "
                    f"Loss: {average_loss:.4f} | "
                    f"Accuracy: {accuracy:.2%}"
                )

        return self

    # predict next token
    def predict_next(self, text, temperature=1.0):
        if temperature <= 0:
            raise ValueError(
                "Temperature must be greater than zero."
            )

        token_ids = self.model.encode(text)

        if not token_ids:
            raise ValueError("Input text cannot be empty.")

        token_ids = token_ids[-self.model.block_size:]

        input_ids = torch.tensor(
            [token_ids],
            dtype=torch.long,
            device=device,
        )

        self.model.eval()

        with torch.no_grad():
            logits = self.model(input_ids)
            next_logits = logits[0, -1] / temperature
            next_logits[self.model.PAD_ID] = -float("inf")

            probabilities = torch.softmax(
                next_logits, dim=-1
            )
            predicted_id = int(
                torch.argmax(probabilities).item()
            )

        return {
            "input": text,
            "predicted_token": self.model.id_to_token[predicted_id],
            "probability": probabilities[predicted_id].item(),
        }

    # plot loss and accuracy charts
    def plot_training_history(self):
        if not self.loss_history:
            raise ValueError("Train the model first using trainer.fit().")
        
        epochs = range(1, len(self.loss_history) + 1)
        # loss and accuracy
        fig, axes = plt.subplots(1, 2, figsize=(16, 5))
        # loss chart
        axes[0].plot(epochs, self.loss_history, label="Training Loss")
        axes[0].set_xlabel("Epoch")
        axes[0].set_ylabel("Cross-Entropy Loss")
        axes[0].set_title("Transformer Training Loss")
        axes[0].grid(True, linestyle="--", alpha=0.3)
        axes[0].legend()
        
        # accuracy chart
        axes[1].plot(epochs,np.array(self.accuracy_history) * 100,label="Training Accuracy",)
        axes[1].set_xlabel("Epoch")
        axes[1].set_ylabel("Accuracy (%)")
        axes[1].set_title("Transformer Training Accuracy")
        axes[1].grid(True, linestyle="--", alpha=0.3)
        axes[1].legend()
        
        fig.text(0.99, 0.01,'\u00A9 Hemant Thapa Copyright 2026', ha='right', va='bottom', fontsize=10, color='black', alpha=0.9)
        fig.tight_layout(rect=[0, 0.05, 1, 1])
        plt.show()
````

````python

# training data
train_texts = [
    """Day after day, day after day,
    We stuck, nor breath nor motion;
    As idle as a painted ship
    Upon a painted ocean.
    Water, water, everywhere,
    And all the boards did shrink;
    Water, water, everywhere,
    Nor any drop to drink.
    The very deep did rot: O Christ!
    That ever this should be!
    Yea, slimy things did crawl with legs
    Upon the slimy sea."""
] * 20
````

````python
# model
model = TinyTransformer(
    texts=train_texts,
    embedding_dim=64,
    num_heads=4,
    num_layers=2,
    feedforward_dim=128,
    block_size=16,
    dropout=0.1,
)

trainer = TransformerTrainer(
    model,
    learning_rate=0.001,
    batch_size=32,
)
````

````python
# train model
trainer.fit(train_texts, epochs=200)
````

````python
# loss and accuracy
trainer.plot_training_history()
````

````python
def predict():
    # predict next token
    user_text = input("Enter text: ")
    result = trainer.predict_next(user_text)
    print("\nInput:", result["input"])
    print("Predicted next token:", result["predicted_token"])
    print("Model probability:", f"{result['probability']:.2%}")
````
